```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Dot product:", np.dot(a, b))
print("Cross product:", np.cross(a, b))
print("Magnitude of a:", np.linalg.norm(a))
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Shape:", A.shape, "Dimensions:", A.ndim)
print("Addition:\n", A + B)
print("Element-wise multiply:\n", A * B)

C = np.arange(24).reshape(2, 3, 4)   # 3D array
print("3D shape:", C.shape)
print("Sum along axis 0:\n", C.sum(axis=0))
```

```python
print("Matrix product:\n", A @ B)
print("Transpose:\n", A.T)
print("Determinant:", np.linalg.det(A))
print("Inverse:\n", np.linalg.inv(A))

eigvals, eigvecs = np.linalg.eig(A)
print("Eigenvalues:", eigvals)
```

```python
# Rotate a vector 90 degrees counterclockwise
theta = np.pi / 2
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
v = np.array([1, 0])
print("Rotated vector:", np.round(R @ v, 2))

# Scaling transformation
S = np.array([[2, 0], [0, 3]])
print("Scaled vector:", S @ np.array([1, 1]))
```

```python
coeff = np.array([[2, 1], [1, 3]])
const = np.array([5, 10])
print("Solution [x, y]:", np.linalg.solve(coeff, const))
```
